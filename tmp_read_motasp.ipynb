```python
# 1. Load Required Libraries
import pandas as pd

# 2. Read the Excel File
excel_path = 'Img/motasp.xlsx'
try:
    xl = pd.ExcelFile(excel_path)
    print('Sheet names:', xl.sheet_names)
except Exception as exc:
    print('Failed to open Excel file:', exc)
    raise

# Load the first sheet by default
sheet_name = xl.sheet_names[0]
df = pd.read_excel(excel_path, sheet_name=sheet_name)
print('Loaded sheet:', sheet_name)

# 3. Inspect DataFrame Structure
print('\nDataFrame shape:', df.shape)
print('Columns:', list(df.columns))
print('\nData types:')
print(df.dtypes)
print('\nMissing values per column:')
print(df.isna().sum())

# 4. Clean and Rename Columns
original_columns = list(df.columns)
clean_columns = [str(col).strip() for col in original_columns]
if clean_columns != original_columns:
    df.columns = clean_columns
    print('\nRenamed columns to remove whitespace:')
    print(df.columns)

# 5. Display Sample Rows
print('\nFirst 10 rows:')
print(df.head(10).to_string(index=False))
print('\nLast 10 rows:')
print(df.tail(10).to_string(index=False))

# 6. Export Processed Data
output_csv = 'tmp_motasp_output.csv'
df.to_csv(output_csv, index=False)
print('\nExported cleaned data to', output_csv)
```
